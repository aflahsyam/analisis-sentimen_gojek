```python
# mengimport library yang dipakai
import time
import pandas as pd
from google_play_scraper import Sort, reviews
```

```python
APP_ID = 'com.gojek.app'
TARGET_PER_SCORE = 7000   # 5 bintang x 7000 = 35.000 data mentah
BATCH = 200
MAX_RETRY = 5

def scrape_by_score(score, target):
    hasil, token, retry = [], None, 0
    while len(hasil) < target:
        try:
            batch, token = reviews(
                APP_ID, lang='id', country='id',
                sort=Sort.NEWEST, count=BATCH,
                filter_score_with=score,
                continuation_token=token
            )
            retry = 0
        except Exception as e:
            retry += 1
            print(f"Error: {e} (retry {retry}/{MAX_RETRY})")
            if retry >= MAX_RETRY:
                break
            time.sleep(5)
            continue
        if not batch:
            break
        hasil.extend(batch)
        print(f"Bintang {score}: {len(hasil)}/{target}")
        if token is None:
            break
        time.sleep(1)
    return hasil[:target]

semua = []
for s in [1, 2, 3, 4, 5]:
    semua.extend(scrape_by_score(s, TARGET_PER_SCORE))
```

```python
df = pd.DataFrame(semua)[['content', 'score']]
df = df.dropna(subset=['content'])
df.to_csv('dataset_gojek_raw.csv', index=False)
print(f"Total data hasil scraping: {len(df)}")
print(df['score'].value_counts().sort_index())
```
